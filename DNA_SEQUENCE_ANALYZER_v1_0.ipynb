```python
# DNA Sequence Analyzer

## Overview

DNA Sequence Analyzer is a Python-based bioinformatics tool developed
to perform basic analysis of DNA sequences.

The project was developed as a learning project to apply programming
concepts to molecular biology and demonstrate how computational tools
can be used to analyse biological sequence data.

## Features

- DNA sequence validation
- Input cleaning
- Nucleotide counting
- GC-content calculation
- DNA complementary strand generation
- DNA-to-RNA transcription
- RNA codon extraction
- RNA-to-protein translation
- Basic sequence reporting

## Technologies

- Python 3
- Google Colab

## Biological concepts

The project applies concepts including:

- DNA nucleotide composition
- Complementary base pairing
- Transcription
- Codons
- Translation
- Genetic code
```

```python
# ============================================
# DNA SEQUENCE ANALYZER v1.0
# ============================================


# --------------------------------------------
# 1. Validate DNA
# --------------------------------------------

def is_valid_dna(dna):
    dna = dna.upper()

    for base in dna:
        if base not in "ATGC":
            return False

    return True


# --------------------------------------------
# 2. Get DNA Complement
# --------------------------------------------

def get_complement(dna):
    complement = ""

    for base in dna:
        if base == "A":
            complement += "T"
        elif base == "T":
            complement += "A"
        elif base == "G":
            complement += "C"
        elif base == "C":
            complement += "G"

    return complement


# --------------------------------------------
# 3. Count Nucleotides
# --------------------------------------------

def nucleotide_counts(dna):
    return {
        "A": dna.count("A"),
        "T": dna.count("T"),
        "G": dna.count("G"),
        "C": dna.count("C")
    }


# --------------------------------------------
# 4. Calculate GC Content
# --------------------------------------------

def gc_content(dna):
    counts = nucleotide_counts(dna)

    return (counts["G"] + counts["C"]) / len(dna) * 100


# --------------------------------------------
# 5. DNA → RNA Transcription
# --------------------------------------------

def transcribe_dna(dna):
    dna = dna.upper()
    dna = dna.replace(" ", "")

    if not is_valid_dna(dna):
        return "Invalid DNA sequence"

    rna = dna.replace("T", "U")

    return rna


# --------------------------------------------
# 6. RNA → Codons
# --------------------------------------------

def get_codons(rna):
    codons = []

    for i in range(0, len(rna), 3):
        codon = rna[i:i+3]
        codons.append(codon)

    return codons


# --------------------------------------------
# 7. Genetic Code
# --------------------------------------------

codon_table = {

    "UUU": "Phenylalanine",
    "UUC": "Phenylalanine",
    "UUA": "Leucine",
    "UUG": "Leucine",

    "UCU": "Serine",
    "UCC": "Serine",
    "UCA": "Serine",
    "UCG": "Serine",

    "UAU": "Tyrosine",
    "UAC": "Tyrosine",
    "UAA": "STOP",
    "UAG": "STOP",

    "UGU": "Cysteine",
    "UGC": "Cysteine",
    "UGA": "STOP",
    "UGG": "Tryptophan",

    "CUU": "Leucine",
    "CUC": "Leucine",
    "CUA": "Leucine",
    "CUG": "Leucine",

    "CCU": "Proline",
    "CCC": "Proline",
    "CCA": "Proline",
    "CCG": "Proline",

    "CAU": "Histidine",
    "CAC": "Histidine",
    "CAA": "Glutamine",
    "CAG": "Glutamine",

    "CGU": "Arginine",
    "CGC": "Arginine",
    "CGA": "Arginine",
    "CGG": "Arginine",

    "AUU": "Isoleucine",
    "AUC": "Isoleucine",
    "AUA": "Isoleucine",
    "AUG": "Methionine",

    "ACU": "Threonine",
    "ACC": "Threonine",
    "ACA": "Threonine",
    "ACG": "Threonine",

    "AAU": "Asparagine",
    "AAC": "Asparagine",
    "AAA": "Lysine",
    "AAG": "Lysine",

    "AGU": "Serine",
    "AGC": "Serine",
    "AGA": "Arginine",
    "AGG": "Arginine",

    "GUU": "Valine",
    "GUC": "Valine",
    "GUA": "Valine",
    "GUG": "Valine",

    "GCU": "Alanine",
    "GCC": "Alanine",
    "GCA": "Alanine",
    "GCG": "Alanine",

    "GAU": "Aspartic acid",
    "GAC": "Aspartic acid",
    "GAA": "Glutamic acid",
    "GAG": "Glutamic acid",

    "GGU": "Glycine",
    "GGC": "Glycine",
    "GGA": "Glycine",
    "GGG": "Glycine"
}


# --------------------------------------------
# 8. RNA → Protein Translation
# --------------------------------------------

def translate_rna(rna):

    rna = rna.upper()
    rna = rna.replace(" ", "")

    # Check that RNA contains only A, U, G and C
    for base in rna:
        if base not in "AUGC":
            return "Invalid RNA sequence"

    codons = get_codons(rna)

    protein = []
    started = False

    for codon in codons:

        # Ignore incomplete codons
        if len(codon) != 3:
            continue

        # Wait for the start codon
        if not started:

            if codon == "AUG":
                started = True
                protein.append("Methionine")

            continue

        # Translate codon
        amino_acid = codon_table[codon]

        # Stop translation
        if amino_acid == "STOP":
            protein.append("STOP")
            break

        protein.append(amino_acid)

    if not started:
        return "No start codon found"

    return protein


# --------------------------------------------
# 9. Complete DNA Analysis
# --------------------------------------------

def analyze_dna(dna):

    dna = dna.upper()
    dna = dna.replace(" ", "")

    if not is_valid_dna(dna):
        print("Invalid DNA sequence")
        return

    length = len(dna)

    counts = nucleotide_counts(dna)

    gc = gc_content(dna)

    complement = get_complement(dna)

    rna = transcribe_dna(dna)

    protein = translate_rna(rna)

    print("================================")
    print("       DNA SEQUENCE ANALYZER")
    print("================================")

    print("Sequence:", dna)
    print("Valid DNA: True")
    print("Length:", length)

    print("\nNUCLEOTIDE COUNTS")
    print("-----------------")
    print("A:", counts["A"])
    print("T:", counts["T"])
    print("G:", counts["G"])
    print("C:", counts["C"])

    print("\nGC CONTENT")
    print("----------")
    print(round(gc, 2), "%")

    print("\nDNA COMPLEMENT")
    print("--------------")
    print(complement)

    print("\nRNA TRANSCRIPT")
    print("--------------")
    print(rna)

    print("\nPROTEIN")
    print("-------")
    print(protein)

    print("================================")
```

```python
dna = "ATGGCTTACTTAAG"

analyze_dna(dna)
```
